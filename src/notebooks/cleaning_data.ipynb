```python
testing_data = df.loc[:,['ticker', 'date', 'adj_open', 'adj_high', 'adj_low', 'adj_close', 'volume']]
testing_data.head(5)
```

```python
# Only include the data for VN30 universe (in 1/2024)
universe = [
    'ACB', 'BCM', 'BID', 'BVH', 'CTG',
    'FPT', 'GAS', 'GVR', 'HDB', 'HPG',
    'MBB', 'MSN', 'MWG', 'PLX', 'POW',
    'SAB', 'SHB', 'SSB', 'SSI', 'STB',
    'TCB', 'TPB', 'VCB', 'VHM', 'VIB',
    'VIC', 'VJC', 'VNM', 'VPB', 'VRE'
]
data = {}
for ticker in universe:
    dat = testing_data[testing_data['ticker'] == ticker]
    dat = dat.loc[:, ['date', 'adj_open', 'adj_high', 'adj_low', 'adj_close', 'volume']]
    dat.set_index('date', inplace=True)
    data[ticker] = dat
```

```python
import pickle

import pandas as pd

with open('/home/supperman/Projects/quant/src/data/trading_date.pickle', 'rb') as f:
    trading_days = pickle.load(f)  # canonical calendar, 2532 days

with open('/home/supperman/Projects/quant/src/data/vn30_ohlcv.pickle', 'rb') as f:
    data = pickle.load(f)


def gap_episodes(missing, calendar):
    """Group missing dates into consecutive-trading-day runs.

    Consecutive positions in the canonical calendar = one episode
    (a suspension stretch), so Friday->Monday gaps count as one episode.
    """
    if len(missing) == 0:
        return []
    pos = calendar.get_indexer(missing)
    episodes, start = [], pos[0]
    for i in range(1, len(pos)):
        if pos[i] != pos[i - 1] + 1:  # break in consecutive positions
            episodes.append((calendar[start], calendar[pos[i - 1]]))
            start = pos[i]
    episodes.append((calendar[start], calendar[pos[-1]]))
    return episodes


rows = []
for ticker, df in data.items():
    idx = df.index
    first, last = idx.min(), idx.max()
    expected = trading_days[(trading_days >= first) & (trading_days <= last)]
    missing = expected.difference(idx)
    episodes = gap_episodes(missing, trading_days)
    rows.append({
        'ticker': ticker,
        'rows': len(idx),
        'first': first.date(),
        'last': last.date(),
        'expected': len(expected),
        'internal_gaps': len(missing),
        'episodes': len(episodes),
        'longest_episode_days': max((e[1] - e[0]).days + 1 for e in episodes) if episodes else 0,
        'extra_dates': len(idx.difference(trading_days)),   # rows on non-trading days
        'duplicates': int(idx.duplicated().sum()),
        'monotonic': bool(idx.is_monotonic_increasing),
    })

res = pd.DataFrame(rows).sort_values('internal_gaps', ascending=False)
print(res.to_string(index=False))

print("\n=== tickers with internal gaps: %d of %d ===" %
      ((res['internal_gaps'] > 0).sum(), len(res)))
print("=== any index anomalies (extra dates / duplicates / non-monotonic)? ===")
anom = res[(res['extra_dates'] > 0) | (res['duplicates'] > 0) | (~res['monotonic'])]
print(anom.to_string(index=False) if len(anom) else "none")

# Show the worst ticker's episodes in detail
worst = res.iloc[0]
ticker = worst['ticker']
missing = trading_days[
    (trading_days >= pd.Timestamp(worst['first'])) &
    (trading_days <= pd.Timestamp(worst['last']))
].difference(data[ticker].index)
eps = gap_episodes(missing, trading_days)
print(f"\n=== longest episodes for {ticker} ({len(eps)} episodes, "
      f"{len(missing)} missing days) ===")
for start, end in sorted(eps, key=lambda e: (e[1] - e[0]).days, reverse=True)[:8]:
    print(f"  {start.date()} -> {end.date()}  ({(end - start).days + 1} calendar days, "
          f"{len(trading_days[(trading_days >= start) & (trading_days <= end)])} trading days)")
```
